```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("netflix_titles.csv")
data
```

Understanding the data

```python
print(data.head(10),"  \n")
print(data.info(),"  \n")
print(data.describe(),"  \n")
print(data.tail(10))
```

Checking and filling the empty data cells

```python
print(data.isnull().sum(),"\n")
data.duplicated().sum()
```

```python
data.fillna("Unknown",inplace = True)
```

```python
data["date_added"] = pd.to_datetime(data["date_added"],errors="coerce")
```

Finding Added month and year

```python
data["Month_Added"] = data["date_added"].dt.month
data
```

```python
data["Year_Added"] =  data['date_added'].dt.year
data
```

Grouping by some fields


```python
print(data['type'].value_counts(),"\n")
print(data['Year_Added'].value_counts(),"\n")
print(data['country'].value_counts().head(10),"\n")#729 rows
```

Visualizing based on Questions :
1. How has the number of titles added changed decades over decades?
2. Movies vs TV Shows — what’s the split?
3. Top 10 genres by count.
4. Top 10 countries producing content.
5. Distribution of movie durations.

```python
# 1
plt.title("Movies Added Over decades on Netflix")
data["decades"] = (data["release_year"]// 10) *10
sns.countplot( x= data["decades"],order=sorted(data['decades'].dropna().unique()))
```

*In* this plot, bucket-2020 , contains only two years 2020 & 2021. Due to dataset cutoff, so no decline can be concluded from it.


```python
# 2
types_Count = data['type'].value_counts()
plt.title("Movies V/S TV shows")
plt.pie(types_Count,labels =["Movies","TV Shows"],colors = ("Green","blue"))
```

```python
# 3
data['genres'] = data['listed_in'].str.split(',\s*')
exploded = data.explode('genres')
genre_counts = exploded['genres'].value_counts()
plt.title("Top Genres")
sns.lineplot(genre_counts.head(10))
plt.xticks(rotation = 45,ha = 'right')
```

```python
# 4
countries = data['country'].value_counts().drop("Unknown", errors='ignore',).head(10)
plt.title("Most content producing countries.")
sns.barplot(x=countries.index, y=countries.values)
plt.xticks(rotation=45, ha='right')
countries
```

```python
# 5
data['duration_minutes'] = data['duration'].where(data['duration'].str.contains('min', na=False)).str.extract('(\d+)').astype(float)
data['duration_seasons'] = data['duration'].where(data['duration'].str.contains('Season', na=False)).str.extract('(\d+)').astype(float)
plt.title("Duration of movies")
sns.histplot(x =data['duration_minutes'])
```

```python
plt.title("Number of seasons in TV Shows")
sns.histplot(x =data['duration_seasons'])
```
